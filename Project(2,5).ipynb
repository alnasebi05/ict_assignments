```python
# Question 2

SUBJECTS = ["Maths", "Science", "English"]

students = []



def get_valid_marks(subject):
    while True:
        marks = int(input(f"Enter marks for {subject} (0-100): "))
        if 0 <= marks <= 100:
            return marks
        print("Invalid marks! Try again.")



def get_grade(subject, marks):
    if subject == "Maths":
        if marks >= 90: return "A+"
        elif marks >= 75: return "A"
        elif marks >= 60: return "B"
        elif marks >= 40: return "C"
        else: return "F"

    elif subject == "Science":
        if marks >= 85: return "A+"
        elif marks >= 70: return "A"
        elif marks >= 55: return "B"
        elif marks >= 40: return "C"
        else: return "F"

    elif subject == "English":
        if marks >= 80: return "A+"
        elif marks >= 65: return "A"
        elif marks >= 50: return "B"
        elif marks >= 40: return "C"
        else: return "F"



def add_student():
    roll = input("Enter Roll Number: ")
    name = input("Enter Name: ")

    marks = {}
    for subject in SUBJECTS:
        marks[subject] = get_valid_marks(subject)

    student = {
        "roll": roll,
        "name": name,
        "marks": marks
    }

    students.append(student)
    print("Student added successfully!\n")



def calculate_average(student):
    total = 0
    for subject in SUBJECTS:
        total += student["marks"][subject]
    return total / len(SUBJECTS)


def get_pass_fail(student):
    for subject in SUBJECTS:
        if student["marks"][subject] < 40:
            return "FAIL"
    return "PASS"



def subject_toppers():
    for subject in SUBJECTS:
        max_marks = -1
        topper = None

        for student in students:
            if student["marks"][subject] > max_marks:
                max_marks = student["marks"][subject]
                topper = student["name"]

        print(f"Topper in {subject}: {topper} ({max_marks})")



def rank_students():
    ranked = students.copy()

    n = len(ranked)
    for i in range(n):
        max_index = i
        for j in range(i + 1, n):
            if calculate_average(ranked[j]) > calculate_average(ranked[max_index]):
                max_index = j

        ranked[i], ranked[max_index] = ranked[max_index], ranked[i]

    print("\nOverall Ranking")
    rank = 1
    for student in ranked:
        print(f"{rank}. {student['name']} - Avg: {calculate_average(student):.2f}")
        rank += 1



def summary_report():
    for student in students:
        print("\nRoll:", student["roll"])
        print("Name:", student["name"])

        for subject in SUBJECTS:
            marks = student["marks"][subject]
            grade = get_grade(subject, marks)
            print(f"{subject}: {marks} | Grade: {grade}")

        print("Average:", calculate_average(student))
        print("Result:", get_pass_fail(student))



def menu():
    while True:
        print("\nStudent Performance Analyzer ")
        print("1. Add Student")
        print("2. View Summary Report")
        print("3. Subject-wise Toppers")
        print("4. Overall Ranking")
        print("5. Exit")

        choice = input("Enter choice: ")

        if choice == "1":
            add_student()
        elif choice == "2":
            summary_report()
        elif choice == "3":
            subject_toppers()
        elif choice == "4":
            rank_students()
        elif choice == "5":
            print("Exiting program")
            break
        else:
            print("Invalid choice!")



menu()
```

```python
#Question 5

rooms = {}
students = {}


def add_room():
    room_no = input("Enter Room Number: ")
    capacity = int(input("Enter Room Capacity: "))
    gender = input("Enter Room Gender (Male/Female/Any): ")

    rooms[room_no] = {
        "capacity": capacity,
        "gender": gender,
        "occupants": []
    }
    print("Room added successfully!")



def register_student():
    roll = input("Enter Roll Number: ")
    name = input("Enter Student Name: ")
    year = int(input("Enter Year (18-50): "))
    gender = input("Enter Gender (Male/Female): ")

    students[roll] = {
        "name": name,
        "year": year,
        "gender": gender,
        "room": None
    }
    print("Student registered successfully!")



def allocate_room():
    roll = input("Enter Student Roll Number: ")

    if roll not in students:
        print("Student not found!")
        return

    if students[roll]["room"] is not None:
        print("Student already has a room!")
        return

    for room_no in rooms:
        room = rooms[room_no]

        if len(room["occupants"]) < room["capacity"]:
            if room["gender"] == "Any" or room["gender"] == students[roll]["gender"]:
                room["occupants"].append(roll)
                students[roll]["room"] = room_no
                print("Room allocated successfully!")
                return

    print("No suitable room available!")



def vacate_room():
    roll = input("Enter Student Roll Number: ")

    if roll not in students or students[roll]["room"] is None:
        print("Student not allocated any room!")
        return

    room_no = students[roll]["room"]
    rooms[room_no]["occupants"].remove(roll)
    students[roll]["room"] = None
    print("Room vacated successfully!")



def room_status():
    for room_no in rooms:
        room = rooms[room_no]
        print("\nRoom:", room_no)
        print("Capacity:", room["capacity"])
        print("Gender:", room["gender"])
        print("Occupants:", len(room["occupants"]))
        for roll in room["occupants"]:
            print(" -", students[roll]["name"])



def allocation_report():
    print("\n Allocation Report")
    for roll in students:
        student = students[roll]
        print("Roll:", roll,
              "| Name:", student["name"],
              "| Room:", student["room"])



def menu():
    while True:
        print("\n Hostel Room Allocation System ")
        print("1. Add Room")
        print("2. Register Student")
        print("3. Allocate Room")
        print("4. Vacate Room")
        print("5. View Room Status")
        print("6. Allocation Report")
        print("7. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            add_room()
        elif choice == "2":
            register_student()
        elif choice == "3":
            allocate_room()
        elif choice == "4":
            vacate_room()
        elif choice == "5":
            room_status()
        elif choice == "6":
            allocation_report()
        elif choice == "7":
            print("Exiting system")
            break
        else:
            print("Invalid choice!")


menu()
```
